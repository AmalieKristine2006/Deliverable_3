#### Task 5:

```python
from Task_4 import Warrior, Mage
from Task_5 import Battle

# Create two different character types
hero = Warrior("Dude")
enemy = Mage("Mr.Wizard")

# Create a battle between the two characters
battle = Battle(hero, enemy)

print("=== Battle starting status ===")
print(hero)
print(enemy)

# Simulate the battle until one character loses
battle.simulate_battle()

print("\n=== Battle result ===")
print(hero)
print(enemy)

if hero.is_alive:
    print(f"{hero.name} wins!")
else:
    print(f"{enemy.name} wins!")
```
